# Why the Gap between simple and log? Volatility (Variance)
log returns will capture the volatility because it measures the geometric growth 

A rough formulation of the difference is:
Avg. log return = Avg. simple return - (Variance/2)
(^ proof with taylor series and rounding higher order terms)

difference between geometric and log returns?
Though similar, geometric is expressed growth on a discrete percentage scale whereas log is expressed on an continuous exponential scale

why use log over pure geometric? log returns are additive

# Calculating returns of portfolio (simple return)

```python
returns_port = (data['Close']/data['Close'].shift(1)) -1
returns_port.head()
```

```python
#now we can weight each out using a numpy array

weights_port = np.array([0.25,0.25,0.25,0.25])
```

```python
#using dot product, we can multiply the weight of each stock by its return

np.dot(returns_port,weights_port)
#because we didnt average out each stocks return, we got this giant array
```

```python
#to fix this, we use the mean to get the average return of each stock
ann_port_return = returns_port.mean() * 250
ann_port_return 
```

```python
#now we can use each stocks average yearly return and dot product it by the weights
ann_port_return_1 = np.dot(ann_port_return,weights_port)
#gives the return of this portfolio yearly based on its weights
```

```python
#we can use different weights to see a different return
weight_2 = np.array([0.15,0.4,0.25,0.20])
```

```python
ann_port_return_2 = np.dot(ann_port_return,weight_2)
ann_port_return_2
```

```python
print(f"Return of Portfolio 1: {str(round(ann_port_return_1,5)*100)}%")
print(f"Return of Portfolio 2: {str(round(ann_port_return_2,7)*100)}%")
```

# Covariance/Correlation for portfolio diversifcation

Correlatio adjusts covriance

# Below is an example of matrix multiplication

```python
A = np.array([[1,3,5],[2,4,6]])
A #2x3 matrix (mxn)
```

```python
B = np.array([[1,4],[2,5],[3,6]])
B #3x2 matrix (nxp)
```

The way you multiply matrice together together is the the inner numbers of the size must match (ie A's columns must match B's Rows) and the outer numbers show the size of the product matrice

in the variance of the portfolio, we first take the dot product of the covariance matrix (2x2) with the weight vector (2x1). This creats a 2x1 column vector for he weights attached to a cariance and covriance term. In NumPy, our weights array we made is neither a 1x2 nor a 2x1 as it is just a 1D array with shape (2,). It only has one axis with a length of 2 rather than separate row and column axes. With no second dimension, NumPy will automatically adjust it depending on where we put the dot product.

For the inner product, itll treat it as a 2x1 column vector, making the multiplcation work. This is becuase we have the weight on the right side. If it was the left side, it would be a 1x2 row vector to make it work. 

np.array([[0.5,0.5]]) would create a 1x2 row vector.

np.array([[0.5],[0.5]]) would create e 2x1 column vector.

Do we even need T with our 1D vector? technically no but it can be good practice with the 2 cases above.

# Calculating diversifiable and undiversifiable risk


```python
weight = np.array([0.5,0.5]) #weights from before
```

```python
weight[0]
```

```python
weight[1] #each weight is accessed separately here
```

A way to get diversifiable risk is to subtract weight annual variances of each stock to the portfolio variance

by putting each stock in another '[]', we turn them into 1x1 matrix (2D arrays)

```python
AAPL_var_a_new = risk_data_returns[['AAPL']].var() * 250
print(AAPL_var_a_new)
AAPL_var_a_new == AAPL_var_a
```

```python
MSFT_var_a_new = risk_data_returns[['MSFT']].var() * 250
print(MSFT_var_a_new)
MSFT_var_a_new == MSFT_var_a
```

```python
#now using the weights from before

non_div_risk = port_var - (weight[0] **2 * AAPL_var_a_new) - (weight[1] ** 2 * MSFT_var_a_new)
non_div_risk
```

The reason the syntax above shows NaN values is because the double brackets creates a Series each. Trying to subtract 2 Series makes the system look to match the values by their index rather than their positions. With "MSFT" not existing in AAPL_var_a_new and "APPL" not existing in MSFT_var_a_new, pandas fills all the missing matches with NaN.

An easy fix is to just use single brackets. Single brackets work because it turns a a 2D table into a 1D list and then into a single number. Double brackets keep the extra dimension intact. This preserves the column bracket making our math break in the end.

we want single numbers as each variable to our equation to get the 1 number in the end.

```python
non_div_risk_actual = port_var - (weight[0] **2 * AAPL_var_a) - (weight[1] ** 2 * MSFT_var_a)
non_div_risk_actual 
```

```python
print(f"Non-Diversifiable Risk: {str(round(non_div_risk_actual*100,3))}%")
```

Diversifiable risk

```python
div_risk = port_var - non_div_risk_actual #non div risk is covariance element
div_risk
```

```python
#or 
div_risk_2 = (weight[0] **2 * AAPL_var_a) + (weight[1] ** 2 * MSFT_var_a)
div_risk_2 
```

```python
div_risk == div_risk_2
```

Why do we have this in variance (squared percentage) vs STD (normal percentage)?

Because risk is split into 2 distinct places. The split only works fully if it is in squared units.

The portfolio variance is the whole thing. Port Var = Div risk + non div risk.

for example 5^2 = 4^2 + 3^2 but 5 ≠ 4 + 3

same way the square root of port var (std of port) ≠ square root of div risk + square root of non div risk

what is valuable here?
we can look at the % split between diversifiable and non diversifiable risk


```python
Non_div_share = non_div_risk_actual / port_var
print(str(round(Non_div_share * 100,3))+'%')
#about 30% of the risk is diversifiable here
```

```python
div_share = div_risk / port_var
print(str(round(div_share * 100,3))+'%')
#about 30% of the risk is diversifiable here
```

```python
total_share = div_share + Non_div_share
print(str(round(total_share * 100,3))+'%')
```

In our math the covariance part of the portfolio variance is what stays. (This is the systematic risk part) while the individual variance terms is what shrinks to 0 as N (number of stocks) grow

by having port_var minus the individual variance parts, we isolate the covariance part (as we see above) to get the market risk. This works for equal weighting.

If it were not equally weighted (ie some stocks having more weight than others), unsystematic risk is driven by the summation of each weight squared (The Herfindahl-Hirschman Index (HHI) Effect). 1 large position can be massively disproportional in inflating unsystematic risk.
