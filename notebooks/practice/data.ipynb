```python
import matplotlib.pyplot as plt
```

```python
PG['simple return'] = (PG['Adj Close'] / PG['Adj Close'].shift(1)) - 1
#(p1/p0) - 1
#shift(1) shifts back the adjusted close 1 day to get the ratio
#.shift(# of lags)
```

```python
PG['simple return']
# because the first one is shifted over, the first date in 1995 is NaN because there was no day before it
#created a column for simple return
```

```python
PG.head()
```

```python
#to plot out the simple returns we created, we can use matplotlib
PG['simple return'].plot(figsize=(8,5)) #(width,height)
#this creates it but wont display it right away^
plt.show() #shows the created plot above
```

```python
#to analyze the average daily returns
avg_returns_d = PG['simple return'].mean()
avg_returns_d
```

```python
#to annualize it
avg_returns_a = PG['simple return'].mean() *250 #trading days
avg_returns_a
```

```python
print(str(round(avg_returns_a, 5) *100) + '%')

#round to 5 decimals (ie 12.263) and then multiply by 100 to get percentage terms
```

Log Returns

# ln(Pt/Pt-1)

```python
PG['log return'] = np.log(PG['Adj Close'] / PG['Adj Close'].shift(1))
PG['log return']
```

```python
PG.head()
#as you can see, the simple and log returns are similiar but do have enough of a difference to be used in different occasions
```

```python
PG['log return'].plot(figsize=(8,5))
plt.show()
#as you can see, the graph looks very similar
```

```python
log_return_d = PG['log return'].mean()
log_return_d
```

```python
log_return_a = PG['log return'].mean() * 250
log_return_a
```

```python
print(str(round(log_return_a,5)*100)+'%')
```

```python
print(f"simple return: {str(round(avg_returns_a, 5) *100) }% vs. log return: {str(round(log_return_a,5)*100)}%")
```

# Why the Gap between simple and log? Volatility (Variance)
log returns will capture the volatility because it measures the geometric growth 

A rough formulation of the difference is:
Avg. log return = Avg. simple return - (Variance/2)
(^ proof with taylor series and rounding higher order terms)

difference between geometric and log returns?
Though similar, geometric is expressed growth on a discrete percentage scale whereas log is expressed on an continuous exponential scale

why use log over pure geometric? log returns are additive
