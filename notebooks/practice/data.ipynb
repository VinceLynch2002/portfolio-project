# Why the Gap between simple and log? Volatility (Variance)
log returns will capture the volatility because it measures the geometric growth 

A rough formulation of the difference is:
Avg. log return = Avg. simple return - (Variance/2)
(^ proof with taylor series and rounding higher order terms)

difference between geometric and log returns?
Though similar, geometric is expressed growth on a discrete percentage scale whereas log is expressed on an continuous exponential scale

why use log over pure geometric? log returns are additive

# Calculating returns of portfolio (simple return)

```python
returns_port = (data['Close']/data['Close'].shift(1)) -1
returns_port.head()
```

```python
#now we can weight each out using a numpy array

weights_port = np.array([0.25,0.25,0.25,0.25])
```

```python
#using dot product, we can multiply the weight of each stock by its return

np.dot(returns_port,weights_port)
#because we didnt average out each stocks return, we got this giant array
```

```python
#to fix this, we use the mean to get the average return of each stock
ann_port_return = returns_port.mean() * 250
ann_port_return 
```

```python
#now we can use each stocks average yearly return and dot product it by the weights
ann_port_return_1 = np.dot(ann_port_return,weights_port)
#gives the return of this portfolio yearly based on its weights
```

```python
#we can use different weights to see a different return
weight_2 = np.array([0.15,0.4,0.25,0.20])
```

```python
ann_port_return_2 = np.dot(ann_port_return,weight_2)
ann_port_return_2
```

```python
print(f"Return of Portfolio 1: {str(round(ann_port_return_1,5)*100)}%")
print(f"Return of Portfolio 2: {str(round(ann_port_return_2,7)*100)}%")
```

# What about risk?

Variance measures the dispersion of a set of data points around their mean value
Standard Deviation is the square root of the variance

```python
#lets look at the risk and return of Microsoft vs Apple

risk_tickers = ['MSFT','AAPL']
risk_data = yf.download(risk_tickers,start='2000-01-01',auto_adjust=True)
risk_data_rfr = yf.download('^IRX',start='2000-01-01',auto_adjust=True)
```

```python
risk_data['Close'].tail()
```

```python
percent_adjusted_rfr = risk_data_rfr['Close']/100
percent_adjusted_rfr
```

```python
#need to use log returns because we are analyzing each separately
risk_data_returns = np.log(risk_data['Close'] / risk_data['Close'].shift(1))
risk_data_returns
```

# MSFT

```python
risk_data_returns['MSFT'].mean()
#average return
```

```python
#average return annualized
risk_data_returns['MSFT'].mean() * 250
```

```python
#standard deviation daily
risk_data_returns['MSFT'].std()
```

```python
#annualized standard deviation
risk_data_returns['MSFT'].std() * (250 ** 0.5) #time is also square rooted
```

# APPL

```python
risk_data_returns['AAPL'].mean()
```

```python
risk_data_returns['AAPL'].mean() * 250
```

```python
risk_data_returns['AAPL'].std()
```

```python
risk_data_returns['AAPL'].std() * (250 ** 0.5)
```

To speed this up, we need to increase the number of dimensions to do the mean and standard deviation of both at the same time



```python
yearly_avg_return = risk_data_returns[['MSFT','AAPL']].mean() * 250
yearly_avg_return
```

```python
yearly_avg_risk = risk_data_returns[['MSFT','AAPL']].std() * (250 ** 0.5)
yearly_avg_risk
```

Above we see that the average yearly returns of apple is much higher but its volatility is also higher (even though not a ton higher)

```python
#sharpe ratio (assuming a hypothetical risk free rate)
(yearly_avg_return - 0.03) / yearly_avg_risk
#higher sharpe ratio indicates a better risk adjusted return
```

```python
yearly_average_rfr = percent_adjusted_rfr.mean()
yearly_average_rfr #long term historical average risk-free rate (it isnt perfect as it doesnt account for volality)
```

```python
rfr_value = yearly_average_rfr.iloc[0] #iloc[0] here extracts the raw number and drops the structure from pandas
#without this, it would be looking for 'AAPl' in the rfr array and vice versa
```

```python
(yearly_avg_return - rfr_value) / yearly_avg_risk
#it shifted upwards from proxy because our rfr is lower
```
