```python
#importing can we done a few ways
#you can import just the module
import math

math.sqrt(100)
```

```python
#or directly sqrt in math (if you just need 1)
from math import sqrt
sqrt(100)
```

```python
#or you can rename the modules or their functions

from math import sqrt as s 

s(100)
```

```python
import math as m
m.sqrt(100)
```

```python
#or if you want everything from a module but dont want to write it all

from math import *
# * imports everything

sqrt(100)
#may not be good if 2 different module both have sqrt
```

```python
help(math)
#shows what it does
```

```python
#can go further 
help(math.sqrt)

# sqrt(x, /)
#     Return the square root of x.

#the forward slash indicates that the arguments before it are postional-only parameters
#ex: math.sqrt(9) --> you can do
#but... math.sqrt(x=9) --> TypeError
```
